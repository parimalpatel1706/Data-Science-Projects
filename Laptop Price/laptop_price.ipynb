```python
import pandas as pd
import joblib
```

```python
df=pd.read_csv(r"C:\Users\PARIMAL\OneDrive\Desktop\Data Science Internship\Projects\Laptop Price\cleaned_laptop_datasets.csv")
```

```python
x=df.drop('Price',axis=1)##independent variable
y=df['Price']## dependent variable
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
## Classifier
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
rfc=RandomForestClassifier()
rfc.fit(x_train,y_train)
print("this is train score:-",rfc.score(x_train,y_train))
print("this is test score:-",rfc.score(x_test,y_test))
model = RandomForestClassifier()
model.fit(x_train,y_train)
```

```python
joblib.dump(model, 'model_compressed.lb', compress=3)
```

```python
prdct1 = rfc.predict(x_test)
print("Predictions:", prdct1)
```
